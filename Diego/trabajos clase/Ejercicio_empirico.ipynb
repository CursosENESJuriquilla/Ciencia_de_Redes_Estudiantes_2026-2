```python
deg_cen =nx.degree_centrality(cc)
max_deg_cen = max(deg_cen, key=deg_cen.get)
```

```python
bet_cen = nx.betweenness_centrality(cc)
max_bet_cen = max(bet_cen, key=bet_cen.get)
```

```python
eig_cen = nx.eigenvector_centrality(cc)
max_eig_cen = max(eig_cen, key = eig_cen.get) 
```

```python
print(max_deg_cen, deg_cen[max_deg_cen])
print(max_bet_cen, bet_cen[max_bet_cen])
print(max_eig_cen, eig_cen[max_eig_cen])
```
