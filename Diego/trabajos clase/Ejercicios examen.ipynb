Considere el grafo  1−2−3−4−1. Calcula el grafo en Networkx, la matriz de adyacencia, el Laplaciano y los eigenvalores.

```python
import networkx as nx
import numpy as np
```

```python
#Dibujo el grafo
G = nx.Graph()
G.add_edges_from([(1,2),(2,3),(3,4),(4,1)])
nx.draw(G)
```

```python
adjancency_m = nx.to_numpy_array(G)
adjancency_m
```

```python
laplacian_m = nx.laplacian_matrix(G).todense()
laplacian_m
```

```python
eigvals = np.linalg.eigvals(laplacian_m)
eigvals
```
