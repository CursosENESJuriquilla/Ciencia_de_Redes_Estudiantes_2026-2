```python
print('Numero de componentes conexas: ',nx.number_connected_components(H))
large = max(nx.connected_components(H), key=len)
size = len(large)
print('Tamaño de componente conexa:', size)
print('Proporcion de nodos en la componente conexa:', size/H.number_of_nodes())
```

```python
degc=nx.betweenness_centrality(S)
deg = sorted(degc, key=degc.get, reverse=True)[:5]
print([(node, degc[node]) for node in deg])
```

```python
degc=nx.degree_centrality(S)
deg = sorted(degc, key=degc.get, reverse=True)[:5]
print([(node, degc[node]) for node in deg])
```
